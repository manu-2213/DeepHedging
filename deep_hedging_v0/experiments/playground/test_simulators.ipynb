```python
from hedging.simulators import HestonSimulator
from experiments.utils.sim_config import (
    load_heston_data,
    EnvConfig,
    PPOConfig,
    TrainingConfig
)
```

```python
params, S0, K, v0 = load_heston_data()
env_cfg = EnvConfig()
ppo_cfg = PPOConfig()
train_cfg = TrainingConfig()
```

```python
r = env_cfg.r
maturity = env_cfg.maturity
trap = env_cfg.trap
num_paths = 10
num_steps = env_cfg.num_steps
history_len = env_cfg.history_len
input_dim = 11
hidden_size = train_cfg.hidden_size
action_dim = 1
transaction_cost = env_cfg.transaction_cost
transaction_fee_rate = env_cfg.transaction_fee_rate
```

```python
simulator = HestonSimulator(
    S0, r, v0, params["theta"], params["rho"], params["kappa"],
    params["sigma"], maturity, num_steps, num_paths
)
```

```python
import time

start = time.time()
simulator.generate_asset_prices()
print(f"Time taken: {time.time() - start}")
```
